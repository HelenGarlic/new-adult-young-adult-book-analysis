```python
output_dir = '/books'  # папка для сохранения

# список файлов
epub_files = ['']
```

```python
import os
import ebooklib
from ebooklib import epub
from bs4 import BeautifulSoup
from pathlib import Path

BLOCKLIST = ['[document]', 'noscript', 'header', 'html', 'meta', 'head', 'input', 'script']

def chapter_to_text(chapter_content: bytes) -> str:
    soup = BeautifulSoup(chapter_content, 'html.parser')
    output, prev = [], ''
    for tag in soup.find_all(string=True):
        if tag.parent.name not in BLOCKLIST and not tag.isspace():
            if not (str(prev).endswith(' ') or str(tag).startswith(' ')):
                output.append('\n\n')
            output.append(str(tag))
            prev = tag
    return ''.join(output)

def convert(epub_path: str, output_dir: str) -> str:
    os.makedirs(output_dir, exist_ok=True)
    book = epub.read_epub(epub_path)
    txt_path = os.path.join(output_dir, Path(epub_path).stem + '.txt')

    with open(txt_path, 'w', encoding='utf-8') as f:
        for item in book.get_items():
            if item.get_type() == ebooklib.ITEM_DOCUMENT:
                f.write(chapter_to_text(item.get_content()) + '\n')

    print(f"✅ Готово: {txt_path}")
    return txt_path
```

```python
for f in epub_files:
    convert(f, output_dir)
```

```python
# Поддерживаемые форматы: mp3, mp4, m4a, wav, ogg, flac, wma
audio_folder = ''
output_txt  = ''
model_size = 'medium'
```

```python
extensions = ('*.mp3', '*.m4a', '*.wav', '*.ogg', '*.flac', '*.wma')
audio_files = []
for ext in extensions:
    audio_files.extend(glob.glob(os.path.join(audio_folder, ext)))

# Сортировка по естественному числовому порядку (глава1, глава2, ..., глава10)
audio_files = natsorted(audio_files)
```

```python
model = whisper.load_model(model_size)

# Папка для промежуточных txt по каждой главе (на случай обрыва)
chapters_dir = str(Path(output_txt).parent / 'chapters_temp')
os.makedirs(chapters_dir, exist_ok=True)

for i, audio_file in enumerate(audio_files, start=1):
    chapter_name = Path(audio_file).stem
    chapter_txt  = os.path.join(chapters_dir, f'{i:03d}_{chapter_name}.txt')

    # Пропускаем уже обработанные главы (возобновление после обрыва)
    if os.path.exists(chapter_txt):
        print(f"⏭️  Глава {i}/{len(audio_files)} уже готова, пропускаем: {chapter_name}")
        continue

    print(f"\n🎧 Глава {i}/{len(audio_files)}: {chapter_name}")
    result = model.transcribe(audio_file, language='ru', verbose=False)

    with open(chapter_txt, 'w', encoding='utf-8') as f:
        f.write(result['text'].strip())

    print(f"✅ Сохранена: {chapter_txt}")

print("\n🔗 Объединяем главы в один файл...")

# Сборка всех глав в правильном порядке
chapter_files = natsorted(glob.glob(os.path.join(chapters_dir, '*.txt')))
with open(output_txt, 'w', encoding='utf-8') as out:
    for chapter_file in chapter_files:
        chapter_name = Path(chapter_file).stem
        with open(chapter_file, 'r', encoding='utf-8') as f:
            out.write(f.read())

print(f"\n📖 Итоговый файл сохранён: {output_txt}")

```

```python
import string
import regex as re
import pandas as pd

from nltk.corpus import stopwords
from nltk.stem import WordNetLemmatizer
from nltk import pos_tag

import pymorphy3 as pm3

# стоп-слова
eng_stopwords = set(stopwords.words('english'))
ru_stopwords  = set(stopwords.words('russian'))

# лемматизаторы
morph = pm3.MorphAnalyzer()       # русский
lemmatizer = WordNetLemmatizer()  # английский


def nltk_pos_tag(word):
    """Простейшее преобразование POS из NLTK в WordNet."""
    tag = pos_tag([word])[0][1][0].upper()
    tag_dict = {'J': 'a', 'N': 'n', 'V': 'v', 'R': 'r'}
    return tag_dict.get(tag, 'n')
```

#### Преобработка текста

```python
def preprocessing(line,
                  lang='en',
                  remove_punctuation=True,
                  remove_stopwords=True,
                  normalisation=True):
    # нижний регистр
    line = line.lower()

    # удаляем разметку глав
    patterns = [
        r'(^|\n)\s*part\s+[a-z0-9ivx]+',
        r'(^|\n)\s*chapter\s+[a-z0-9ivx]+',
        r'(^|\n)\s*глава\s+[а-яё0-9ivx]+',
        r'(^|\n)\s*часть\s+[а-яё0-9ivx]+',
    ]
    for pat in patterns:
        line = re.sub(pat, r'\1 ', line)

    # переносы → пробелы
    line = re.sub('\n\t', ' ', line)

    # оставляем только рус/англ буквы, цифры и пунктуацию
    line = re.sub(r'[^а-яёa-z0-9\s{}]'.format(string.punctuation), '', line)

    # убрать пунктуацию
    if remove_punctuation:
        line = re.sub(r"[{}]".format(string.punctuation), " ", line)

    # токенизация по пробелам
    words = [x for x in re.split(r'\s+', line) if x]

    # удаление стоп-слов
    if remove_stopwords:
        stop_words = eng_stopwords if lang == 'en' else ru_stopwords
        words = [x for x in words if x not in stop_words]

    # лемматизация
    if normalisation:
        lemmas = []
        if lang == 'ru':
            for x in words:
                p = morph.parse(x)[0]
                lemmas.append(p.normal_form)
        elif lang == 'en':
            for x in words:
                pos = nltk_pos_tag(x)
                lemmas.append(lemmatizer.lemmatize(x, pos))
        else:
            lemmas = words
    else:
        lemmas = words

    return ' '.join(lemmas)
```

```python
def words_in_book(b_id):
    b_text = book_df[book_df['id'] == b_id]['text'].astype(str).iloc[0]
    words = re.sub(r'[.!?]', ' ', b_text)
    return words.split()
```

```python
def word_len_stat(b_id):
    words = words_in_book(b_id)
    lengths = np.array([len(w) for w in words])
    return (
        len(words),                # n_words
        float(np.mean(lengths)),   # mean
        int(np.min(lengths)),      # min
        int(np.max(lengths)),      # max
        float(np.median(lengths)), # median
        float(np.std(lengths)),    # std
    )
```

```python
def count_sent_in_book(row):
    text = str(row['text'])
    pattern = r'[.!?]+[\s\n]+'
    sentences = [s for s in re.split(pattern, text) if s.strip()] # убирает пустые строки
    return len(sentences)
```

```python
VOWELS = "aeiouy"

def count_syllables_word(word: str) -> int:
    word = word.lower()
    word = re.sub(r'[^a-z]', '', word)
    if not word:
        return 0

    count = 0
    if word[0] in VOWELS:
        count += 1
    for i in range(1, len(word)):
        if word[i] in VOWELS and word[i-1] not in VOWELS:
            count += 1
    if word.endswith("e"):
        count -= 1
    if word.endswith("le") and len(word) > 2 and word[-3] not in VOWELS:
        count += 1
    if count <= 0:
        count = 1
    return count


def count_syllables_text(text: str) -> int:
    if pd.isna(text):
        return 0
    words = re.findall(r"[A-Za-z']+", str(text))
    return sum(count_syllables_word(w) for w in words)
```

```python
# подсчет слогов в русском слове / предложении
VOWELS_RU = "аеёиоуыэюя"
def count_syllables_word_ru(word: str) -> int:
    word = word.lower()
    # оставляем только русские буквы
    word = re.sub(r'[^а-яё]', '', word)

    if not word:
        return 0

    # число слогов = количество гласных
    count = sum(1 for ch in word if ch in VOWELS_RU)

    if count <= 0:
        count = 1
    return count

def count_syllables_text_ru(text: str) -> int:
    if pd.isna(text):
        return 0
    # слова как последовательности русских букв
    words = re.findall(r"[А-Яа-яЁё]+", str(text))
    return sum(count_syllables_word_ru(w) for w in words)

```

```python
# ─── Вспомогательные функции для русского ───────────────────────────────────

VOWELS_RU = re.compile(r'[аеёиоуыьъэюяАЕЁИОУЫЬЪЭЮЯ]')
SENT_SPLIT_RU = re.compile(r'[.!?]+')
WORD_RE = re.compile(r'\b\w+\b')

def count_syllables_ru(word: str) -> int:
    return len(VOWELS_RU.findall(word))

def avg_syllables_per_word_ru(text: str) -> float:
    words = WORD_RE.findall(text)
    if not words:
        return 0.0
    return sum(count_syllables_ru(w) for w in words) / len(words)

def flesh_ru(text: str) -> float:
    """Адаптированный индекс Флеша для русского."""
    sents = [s.strip() for s in SENT_SPLIT_RU.split(text) if s.strip()]
    words = WORD_RE.findall(text)
    if not sents or not words:
        return 0.0
    asl = len(words) / len(sents)          # средняя длина предложения
    asw = avg_syllables_per_word_ru(text)  # среднее кол-во слогов
    return 206.835 - 1.3 * asl - 60.1 * asw

def smog_ru(text: str) -> int:
    """SMOG grade для русского (приближение)."""
    sents = [s.strip() for s in SENT_SPLIT_RU.split(text) if s.strip()]
    words = WORD_RE.findall(text)
    if len(sents) < 3:
        return 0
    polysyl = sum(1 for w in words if count_syllables_ru(w) >= 3)
    return int(3 + np.sqrt(polysyl * (30 / len(sents))))

# ─── Основная функция ────────────────────────────────────────────────────────

def compute_readability(text: str, lang: str = 'en') -> dict:
    text = str(text)
    if lang == 'en':
        return {
            'flesch_reading_ease':   textstat.flesch_reading_ease(text),
            'flesch_kincaid_grade':  textstat.flesch_kincaid_grade(text),
            'gunning_fog':           textstat.gunning_fog(text),
            'smog_index':            textstat.smog_index(text),
            'automated_readability': textstat.automated_readability_index(text),
            'dale_chall':            textstat.dale_chall_readability_score(text),
        }
    else:  # ru
        return {
            'flesch_ru':  flesh_ru(text),
            'smog_ru':    smog_ru(text),
            'avg_syl_per_word': avg_syllables_per_word_ru(text),
            'avg_sent_len': len(WORD_RE.findall(text)) /
                            max(1, len([s for s in SENT_SPLIT_RU.split(text) if s.strip()])),
        }
```

```python
def clean_for_readability(text: str) -> str:
    """
    Очищает текст от технической разметки, заголовков глав/частей,
    лишних переносов и пробелов для корректного анализа читаемости.
    """
    text = str(text)

    # 1. Убираем заголовки глав и частей (EN + RU, регистронезависимо)
    # chapter_patterns = [
    #     r'(^|\n)\s*part\s+[a-z0-9ivx]+[^\n]*',
    #     r'(^|\n)\s*chapter\s+[a-z0-9ivx]+[^\n]*',
    #     r'(^|\n)\s*глава\s+[а-яё0-9ivx]+[^\n]*',
    #     r'(^|\n)\s*часть\s+[а-яё0-9ivx]+[^\n]*',
    # ]

    chapter_patterns = [
        r'\b(part|chapter)\s*[a-z0-9ivxlcdm]*\.?',
        r'\b(prologue|epilogue)\.?',
        r'\b(глава|часть|книга|раздел)\s*[а-яёa-z0-9]*\.?',
        r'\b(эпилог|пролог)\.?',
    ]
    for pat in chapter_patterns:
        text = re.sub(pat, ' ', text, flags=re.IGNORECASE)

    # 2. Убираем HTML/XML теги
    text = re.sub(r'<[^>]+>', ' ', text)

    # 3. Убираем URL
    text = re.sub(r'https?://\S+', ' ', text)

    # 4. Табуляции → пробел
    text = re.sub(r'\t', ' ', text)

    # 5. Переносы строк: одиночный \n → пробел (продолжение предложения),
    #    двойной \n\n → точка + пробел (граница абзаца, важно для читаемости)
    text = re.sub(r'\n{2,}', '. ', text)
    text = re.sub(r'\n', ' ', text)

    # 6. Специальные пробельные символы Unicode (\xa0, \u200b и др.) → пробел
    text = re.sub(r'[\xa0\u200b\u200c\u200d\ufeff]', ' ', text)

    # 7. Множественные пробелы → один
    text = re.sub(r' {2,}', ' ', text)

    return text.strip()

```

### TTR — Type-Token Ratio

```python
def compute_ttr(text: str) -> float:
    words = re.findall(r'\b\w+\b', str(text).lower())
    if not words:
        return 0.0
    return len(set(words)) / len(words)

# Вариант MATTR (скользящее окно, устойчивее к длине текста)
def compute_mattr(text: str, window: int = 500) -> float:
    words = re.findall(r'\b\w+\b', str(text).lower())
    if len(words) < window:
        return compute_ttr(text)
    ttrs = [
        len(set(words[i:i+window])) / window
        for i in range(0, len(words) - window + 1, window)
    ]
    return float(np.mean(ttrs))
```

```python
# ─── Universal Dependencies UPOS (русский, Natasha) ──────────────────────────
def pos_stats_ru(text: str, sample_chars: int = 100_000) -> dict:
    doc = Doc(str(text))
    doc.segment(segmenter)
    doc.tag_morph(morph_tagger)
    counts = {}
    for token in doc.tokens:
        pos = token.pos
        if pos:
            counts[pos] = counts.get(pos, 0) + 1
    return counts
```

```python
# ─── POS-теги Penn Treebank (английский, spaCy) ──────────────────────────────
def pos_stats_en(text: str, sample_chars: int = 100_000) -> dict:
    doc = nlp_en(str(text))
    counts = {}
    for token in doc:
        if not token.is_space and not token.is_punct:
            counts[token.pos_] = counts.get(token.pos_, 0) + 1
    return counts
```

```python
def get_top10(text, lang='en'):
    stops = stop_en if lang == 'en' else stop_ru
    words = re.findall(r'\b[a-zA-Zа-яёА-ЯЁ]+\b', text.lower())
    words = [w for w in words if w not in stops]
    return Counter(words).most_common(10)
```

```python
def plot_book_distribution(df: pd.DataFrame, col: str, xlabel: str,
                           title: str, lang_col: str = 'language',
                           en_val: str = 'en', ru_val: str = 'ru'):
    """
    Горизонтальные столбцы для каждой книги из одного датафрейма.
    EN — синие (левая панель), RU — красные (правая панель).
    """
    en_df = df[df[lang_col] == en_val].copy()
    ru_df = df[df[lang_col] == ru_val].copy()

    fig, axes = plt.subplots(1, 2, figsize=(16, 5))

    for ax, sub_df, lang, color in zip(
        axes,
        [en_df, ru_df],
        ['EN', 'RU'],
        ['steelblue', 'tomato']
    ):
        sub_df = sub_df.sort_values(col)
        labels = sub_df['title'].values
        values = sub_df[col].values

        bars = ax.barh(labels, values, color=color)
        for bar in bars:
            w = bar.get_width()
            ax.text(w * 1.01, bar.get_y() + bar.get_height() / 2,
                    f'{int(w):,}', va='center', fontsize=9)

        ax.set_title(f'{title} ({lang})', fontsize=13)
        ax.set_xlabel(xlabel)
        ax.grid(axis='x', linestyle='--', alpha=0.6)

    plt.tight_layout()
    plt.show()
```

```python
def plot_pos_pie(pos_dict: dict, title: str, label_map: dict = None,
                threshold: float = 0.03, pct_threshold: float = 0.05):
    """
    pct_threshold — минимальная доля для отображения % внутри сектора.
                    Для маленьких секторов процент скрывается.
    """
    if label_map is None:
        label_map = {}

    total = sum(pos_dict.values())
    if total == 0:
        print(f"[{title}] Пустой словарь, пропускаем.")
        return

    renamed = {}
    other_sum = 0
    other_details = {}

    for pos, count in pos_dict.items():
        name = label_map.get(pos, pos)
        share = count / total
        if share < threshold:
            other_sum += count
            other_details[name] = round(share * 100, 1)
        else:
            renamed[name] = renamed.get(name, 0) + count

    if other_sum > 0:
        renamed['Другое'] = other_sum

    if not renamed:
        print(f"[{title}] Все категории < {threshold*100}%, снизь threshold.")
        return

    sorted_pos = dict(sorted(renamed.items(), key=lambda x: x[1], reverse=True))
    values = list(sorted_pos.values())
    labels = list(sorted_pos.keys())
    colors = plt.cm.Paired(np.linspace(0, 1, len(sorted_pos)))

    fig, ax = plt.subplots(figsize=(10, 7))

    # Показываем % только для секторов выше pct_threshold
    def autopct_fn(pct):
        return f'{pct:.1f}%' if pct / 100 >= pct_threshold else ''

    wedges, _, autotexts = ax.pie(
        values,
        labels=None,
        autopct=autopct_fn,          # ← умная функция вместо строки формата
        startangle=90,
        colors=colors,
        textprops={'fontsize': 11},
        pctdistance=0.75,            # ← подписи ближе к центру
    )

    for at in autotexts:
        at.set_color('white')
        at.set_fontweight('bold')

    # Добавляем % в легенду
    legend_labels = [
        f"{name} ({v/total*100:.1f}%)" for name, v in sorted_pos.items()
    ]
    ax.legend(
        wedges, legend_labels,
        title='Части речи',
        loc='center left',
        bbox_to_anchor=(1, 0, 0.5, 1),
        fontsize=10
    )

    ax.set_title(title, fontsize=15)
    ax.axis('equal')
    plt.tight_layout()
    plt.show()

    if other_details:
        print(f"\nКатегория 'Другое' ({title}):")
        for name, pct in sorted(other_details.items(), key=lambda x: -x[1]):
            print(f"  {name}: {pct}%")
```

```python
import re

def clean_text_ru(text):
    # Убрать заголовки глав (русские варианты)
    text = re.sub(r'(Часть\s+\w+|Глава\s+\w+|Пролог|Эпилог)', '', text, flags=re.IGNORECASE)
    text = text.replace('\r', ' ')
    text = text.replace('\n', ' ')
    text = text.replace('\t', ' ')
    text = re.sub(r'\s+', ' ', text)
    return text.strip()
```

## Контекстный анализ лексем *love* и *любовь/любить* в двуязычном корпусе романтической прозы

```python
import spacy
import pymorphy3
from collections import Counter, defaultdict
import pandas as pd
import matplotlib.pyplot as plt
import matplotlib
from matplotlib.gridspec import GridSpec
import re

matplotlib.rcParams['font.family'] = 'DejaVu Sans'

nlp_en = spacy.load('en_core_web_sm')
morph = pymorphy3.MorphAnalyzer()
```

```python
WINDOW = 3  # окно ±3 слова

def get_love_contexts_en(texts: list, window: int = WINDOW):
    """
    Извлекает контекстные окна вокруг love-NOUN и love-VERB из EN текстов.
    texts — список строк (тексты книг)
    Возвращает два списка соседних лемм: для noun и для verb.
    """
    noun_context = []
    verb_context = []

    EN_STOP = {'not', 'i', 'd', 're', 've', 'm', 's',
                'the', 'a', 'an', 'it', 'be', 'do',
                'that', 'this', 'so', 'just', 'get'
                }

    for text in texts:
        doc = nlp_en(str(text))
        tokens = list(doc)
        for i, token in enumerate(tokens):
            if token.lemma_.lower() != 'love':
                continue
            # Соседи в окне
            left  = tokens[max(0, i - window): i]
            right = tokens[i + 1: i + 1 + window]
            neighbors = [t.lemma_.lower() for t in left + right
                         if not t.is_punct and not t.is_space
                         and t.lemma_.lower() not in EN_STOP
                         and len(t.lemma_) > 1
                         and t.lemma_.lower() not in ('love',)]

            if token.pos_ == 'NOUN':
                noun_context.extend(neighbors)
            elif token.pos_ == 'VERB':
                verb_context.extend(neighbors)

    return noun_context, verb_context


def get_love_contexts_ru(texts: list, window: int = WINDOW):
    """
    Извлекает контекстные окна вокруг любовь (NOUN) и любить (VERB) из RU текстов.
    Использует pymorphy3 для лемматизации.
    """
    noun_context = []
    verb_context = []

    RU_STOP = {'и', 'в', 'на', 'с', 'по', 'к', 'за', 'из', 'не',
                 'то', 'а', 'но', 'же', 'бы', 'как', 'что', 'это',
                 'он', 'она', 'они', 'мы', 'я', 'ты', 'вы', 'его',
                 'ее', 'её', 'их', 'мне', 'меня', 'себя', 'себе',
                 'который', 'кто', 'мой', 'твой', 'свой', 'тот',
                'такой', 'весь', 'быть'}

    for text in texts:
        # Разбиваем на токены (слова)
        raw_tokens = re.findall(r'[а-яёА-ЯЁa-zA-Z]+', str(text))
        # Лемматизируем все токены
        lemmas = []
        pos_tags = []
        for w in raw_tokens:
            parsed = morph.parse(w.lower())[0]
            lemmas.append(parsed.normal_form)
            pos_tags.append(parsed.tag.POS)

        for i, (lemma, pos) in enumerate(zip(lemmas, pos_tags)):
            target_lemma = None
            target_type  = None

            if lemma == 'любовь':
                target_lemma = 'любовь'
                target_type  = 'noun'
            elif lemma == 'любить':
                target_lemma = 'любить'
                target_type  = 'verb'

            if target_lemma is None:
                continue

            left  = lemmas[max(0, i - window): i]
            right = lemmas[i + 1: i + 1 + window]
            neighbors = [l for l in left + right
                         if l not in RU_STOP and len(l) > 2
                         and l not in ('любовь', 'любить')]

            if target_type == 'noun':
                noun_context.extend(neighbors)
            else:
                verb_context.extend(neighbors)

    return noun_context, verb_context


```

```python
en_texts = [en_analyze_df[en_analyze_df['id'] == i]['text_clean'].values[0] for i in en_analyze_df['id'].unique()]
ru_texts = [ru_analyze_df[ru_analyze_df['id'] == i]['text_clean'].values[0] for i in ru_analyze_df['id'].unique()]

en_noun_ctx, en_verb_ctx = get_love_contexts_en(en_texts)
ru_noun_ctx, ru_verb_ctx = get_love_contexts_ru(ru_texts)

```

```python
TOP_N = 15

en_noun_freq = Counter(en_noun_ctx).most_common(TOP_N)
en_verb_freq = Counter(en_verb_ctx).most_common(TOP_N)
ru_noun_freq = Counter(ru_noun_ctx).most_common(TOP_N)
ru_verb_freq = Counter(ru_verb_ctx).most_common(TOP_N)

# Вывод таблиц
for label, freq in [('EN love-NOUN', en_noun_freq),
                     ('EN love-VERB', en_verb_freq),
                     ('RU любовь (NOUN)', ru_noun_freq),
                     ('RU любить (VERB)', ru_verb_freq)]:
    df = pd.DataFrame(freq, columns=['word', 'freq'])
    print(f"\n=== {label} ===")
    print(df.to_string(index=False))

```

```python
fig = plt.figure(figsize=(16, 12))
fig.suptitle('Контекстное окружение лексемы LOVE/ЛЮБОВЬ в зависимости\nот частеречной функции (топ-15 соседних слов)',
             fontsize=14, fontweight='bold', y=0.98)

gs = GridSpec(2, 2, figure=fig, hspace=0.5, wspace=0.4)

panels = [
    (en_noun_freq, 'EN: love-NOUN\n(контекст существительного)', '#4C72B0', gs[0, 0]),
    (en_verb_freq, 'EN: love-VERB\n(контекст глагола)',           '#DD8452', gs[0, 1]),
    (ru_noun_freq, 'RU: любовь-NOUN\n(контекст существительного)','#55A868', gs[1, 0]),
    (ru_verb_freq, 'RU: любить-VERB\n(контекст глагола)',         '#C44E52', gs[1, 1]),
]

for freq_data, title, color, grid_pos in panels:
    ax = fig.add_subplot(grid_pos)
    if not freq_data:
        ax.text(0.5, 0.5, 'Нет данных', ha='center', va='center', transform=ax.transAxes)
        ax.set_title(title, fontsize=11, fontweight='bold')
        continue
    words  = [w for w, _ in freq_data][::-1]
    counts = [c for _, c in freq_data][::-1]
    bars = ax.barh(words, counts, color=color, alpha=0.85)
    ax.set_title(title, fontsize=11, fontweight='bold')
    ax.set_xlabel('Частота', fontsize=9)
    # Подписи значений
    for bar, val in zip(bars, counts):
        ax.text(val + 0.1, bar.get_y() + bar.get_height()/2,
                str(val), va='center', fontsize=8)
    ax.spines['top'].set_visible(False)
    ax.spines['right'].set_visible(False)

plt.show()
```
